````python
import sqlite3
import openai
import pandas as pd
from google.colab import userdata

# 1. Securely read your new API key from Colab Secrets
try:
    api_key = userdata.get('OPENAI_API_KEY')
    client = openai.OpenAI(api_key=api_key)
except Exception as e:
    print("⚠️ SECRET ERROR: Please make sure 'OPENAI_API_KEY' is added to your Colab Key icon panel and 'Notebook access' is turned ON!")

# 2. This function extracts the exact table structure so the AI doesn't guess column names
def get_db_structure():
    conn = sqlite3.connect("chinook.db")
    cursor = conn.cursor()
    cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
    tables = cursor.fetchall()

    structure = ""
    for table in tables:
        table_name = table[0]
        if table_name.startswith("sqlite_"): continue
        structure += f"\nTable: {table_name}\nColumns:\n"
        cursor.execute(f"PRAGMA table_info({table_name});")
        for col in cursor.fetchall():
            structure += f"  - {col[1]}\n"
    conn.close()
    return structure

# 3. This is the main function that converts your english into data results
def ask_database(english_question):
    db_layout = get_db_structure()

    system_instructions = f"""
    You are an expert SQL generator. Read the database layout below and turn the user's question into a valid SQLite query.
    Output ONLY the raw SQL code. Do not include markdown formatting, code blocks (like ```sql), or explanations.

    Database Layout:
    {db_layout}
    """

    # Send instructions to OpenAI
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": system_instructions},
            {"role": "user", "content": english_question}
        ],
        temperature=0.0
    )

    generated_sql = response.choices[0].message.content.strip()

    # Clean up accidental markdown backticks if the AI provides them anyway
    if generated_sql.startswith("```"):
        generated_sql = generated_sql.split("\n", 1)[1].rsplit("\n", 1)[0].strip()
    if generated_sql.lower().startswith("sql"):
        generated_sql = generated_sql[3:].strip()

    print(f"🤖 AI Generated SQL Code:\n{generated_sql}\n")
    print("📊 Database Answer Table:")

    # Run the query against the database file and print it as a nice visual spreadsheet
    conn = sqlite3.connect("chinook.db")
    df = pd.read_sql_query(generated_sql, conn)
    conn.close()

    return df
````

````python
# 1. Install Google GenAI package
!pip install -q google-genai

import sqlite3
import pandas as pd
from google.colab import userdata
from google import genai

# 2. Connect to free Gemini Client
try:
    gemini_key = userdata.get('GEMINI_API_KEY')
    client = genai.Client(api_key=gemini_key)
except Exception as e:
    print("⚠️ Please make sure 'GEMINI_API_KEY' is added to your Colab Key icon panel!")

def get_db_structure():
    conn = sqlite3.connect("chinook.db")
    cursor = conn.cursor()
    cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
    tables = cursor.fetchall()
    structure = ""
    for table in tables:
        table_name = table[0]
        if table_name.startswith("sqlite_"): continue
        structure += f"\nTable: {table_name}\nColumns:\n"
        cursor.execute(f"PRAGMA table_info({table_name});")
        for col in cursor.fetchall():
            structure += f"  - {col[1]}\n"
    conn.close()
    return structure

def ask_database(english_question):
    db_layout = get_db_structure()

    system_instructions = f"""
    You are an expert SQL generator. Read the database layout below and turn the user's question into a valid SQLite query.
    Output ONLY the raw SQL code. Do not include markdown formatting, code blocks (like ```sql), or explanations.

    Database Layout:
    {db_layout}
    """

    # Generate content using free gemini-2.5-flash
    response = client.models.generate_content(
        model='gemini-2.5-flash',
        contents=english_question,
        config={'system_instruction': system_instructions, 'temperature': 0.0}
    )

    generated_sql = response.text.strip()

    if generated_sql.startswith("```"):
        generated_sql = generated_sql.split("\n", 1)[1].rsplit("\n", 1)[0].strip()
    if generated_sql.lower().startswith("sql"):
        generated_sql = generated_sql[3:].strip()

    print(f"🤖 AI Generated SQL Code:\n{generated_sql}\n")
    print("📊 Database Answer Table:")

    conn = sqlite3.connect("chinook.db")
    df = pd.read_sql_query(generated_sql, conn)
    conn.close()

    return df
````

````python
import sqlite3

conn = sqlite3.connect("chinook.db")
cursor = conn.cursor()
cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
tables = cursor.fetchall()

print("🗂️ COMPLETE CHINOOK DATABASE MAP:")
print("=" * 40)

for table in tables:
    table_name = table[0]
    if table_name.startswith("sqlite_"): continue

    print(f"\n📦 Table: {table_name}")
    cursor.execute(f"PRAGMA table_info({table_name});")
    cols = cursor.fetchall()
    col_names = [f"{col[1]} ({col[2]})" for col in cols]
    print(f"   Columns: {', '.join(col_names)}")

conn.close()
````
